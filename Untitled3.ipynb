```python

from IPython.display import HTML, display
import html

game_html = r"""
<!DOCTYPE html>
<html>
<head>
<meta charset="UTF-8">
<meta name="viewport" content="width=device-width, initial-scale=1">
<style>
* { box-sizing:border-box; }
body {
    margin:0;
    padding:20px 8px;
    background:#dce3e8;
    font-family:Arial,sans-serif;
}
.phone {
    width: min(100%, 390px);
    margin:auto;
    padding:22px 20px 25px;
    border-radius:48px 48px 55px 55px;
    background:linear-gradient(135deg,#46566a,#253448 55%,#182536);
    box-shadow:inset 3px 3px 5px #8190a1,
               inset -4px -5px 8px #101a28,
               0 18px 35px #0005;
    border:3px solid #182536;
}
.brand {
    color:#e3e9ee;
    text-align:center;
    font-size:25px;
    font-weight:bold;
    letter-spacing:3px;
    margin:0 0 17px;
    text-shadow:1px 2px 2px #111;
}
.brand small {
    font-size:9px;
    letter-spacing:2px;
    display:block;
    margin-top:3px;
    color:#aab7c5;
}
.screen-frame {
    padding:12px;
    border-radius:17px;
    background:#18252c;
    box-shadow:inset 0 3px 8px #000;
}
.screen {
    position:relative;
    background:#b9cda0;
    border:3px solid #607b4b;
    border-radius:3px;
    overflow:hidden;
    box-shadow:inset 0 0 15px #617b48;
}
.screen::after {
    content:"";
    pointer-events:none;
    position:absolute;
    inset:0;
    background:repeating-linear-gradient(
        transparent 0px,transparent 3px,
        #38552b12 4px
    );
}
.topbar {
    height:27px;
    padding:5px 8px;
    display:flex;
    justify-content:space-between;
    color:#263b21;
    font-family:monospace;
    font-weight:bold;
    font-size:12px;
}
canvas {
    display:block;
    width:100%;
    height:auto;
    image-rendering:pixelated;
}
.screen-footer {
    display:flex;
    justify-content:space-between;
    padding:5px 8px;
    color:#263b21;
    font:bold 11px monospace;
}
.overlay {
    position:absolute;
    inset:27px 0 25px;
    background:#b9cda0ed;
    display:flex;
    flex-direction:column;
    justify-content:center;
    align-items:center;
    text-align:center;
    color:#20351e;
    font-family:monospace;
    z-index:2;
    padding:12px;
}
.overlay h2 {
    font-size:23px;
    margin:4px;
    letter-spacing:2px;
}
.overlay p { font-size:12px; line-height:1.7; }
.pixel-btn {
    background:#34483b;
    color:#d5e5bb;
    border:2px solid #82966c;
    padding:9px 19px;
    font:bold 12px monospace;
    cursor:pointer;
    border-radius:4px;
    margin:5px;
    box-shadow:0 3px 0 #17281d;
}
.pixel-btn:active { transform:translateY(2px);box-shadow:none; }
.controls {
    display:flex;
    flex-direction:column;
    align-items:center;
    margin-top:22px;
    gap:5px;
}
.control-row { display:flex;gap:7px; }
.arrow {
    width:53px;
    height:43px;
    border:1px solid #8191a2;
    border-radius:10px;
    background:linear-gradient(#697c90,#34475c);
    color:white;
    font-size:22px;
    cursor:pointer;
    box-shadow:0 4px 0 #172638;
}
.arrow:active { transform:translateY(3px);box-shadow:none; }
.nav-area {
    display:flex;
    justify-content:space-around;
    align-items:center;
    margin:23px 0 15px;
}
.nav-circle {
    width:88px;height:55px;
    background:linear-gradient(#526579,#253548);
    border-radius:35px;
    border:2px solid #8290a0;
    color:#e3e8ee;
    display:flex;
    justify-content:center;
    align-items:center;
    font-size:12px;
    font-weight:bold;
    cursor:pointer;
    box-shadow:0 4px 0 #152232;
}
.nav-center {
    width:50px;height:50px;
    border-radius:50%;
    background:linear-gradient(#a2adba,#536579);
    border:3px solid #253448;
    display:flex;align-items:center;justify-content:center;
    color:#172536;font-weight:bold;font-size:11px;
}
.keypad {
    display:grid;
    grid-template-columns:repeat(3,1fr);
    gap:9px;
    padding:5px 15px;
}
.key {
    height:40px;
    border-radius:12px;
    background:linear-gradient(#61768c,#2c3c50);
    color:#eef1f5;
    text-align:center;
    display:flex;
    justify-content:center;
    align-items:center;
    font-size:16px;
    font-weight:bold;
    border:1px solid #8493a2;
    box-shadow:0 3px 0 #172435;
}
.key small { font-size:9px;margin-left:5px;color:#c0cbd5; }
.phone-bottom {
    text-align:center;
    color:#a9b7c5;
    font-size:9px;
    letter-spacing:3px;
    margin-top:17px;
}
.actions {
    display:flex;
    justify-content:center;
    flex-wrap:wrap;
    gap:7px;
    margin-top:15px;
}
.instructions {
    text-align:center;
    color:#d5e0e9;
    font:10px monospace;
    line-height:1.8;
    margin-top:13px;
}
</style>
</head>
<body>

<div class="phone">
    <div class="brand">NOKIA<small>3310 CLASSIC MOBILE</small></div>

    <div class="screen-frame">
        <div class="screen">
            <div class="topbar">
                <span>◈ NOKIA</span>
                <span id="status">READY</span>
            </div>

            <canvas id="game" width="320" height="320"></canvas>

            <div class="screen-footer">
                <span>SCORE: <span id="score">0</span></span>
                <span>BEST: <span id="best">0</span></span>
            </div>

            <div class="overlay" id="overlay">
                <h2>SNAKE</h2>
                <h2>XENZIA</h2>
                <p>
                    THE CLASSIC NOKIA GAME<br><br>
                    EAT FOOD TO GROW!<br>
                    AVOID WALLS & YOUR TAIL<br><br>
                    ARROWS / WASD TO MOVE<br>
                    SPACE TO PAUSE
                </p>
                <button class="pixel-btn" onclick="startGame()">▶ START GAME</button>
            </div>
        </div>
    </div>

    <div class="nav-area">
        <div class="nav-circle" onclick="toggleSound()" id="soundBtn">SOUND ON</div>
        <div class="nav-center">OK</div>
        <div class="nav-circle" onclick="togglePause()">PAUSE</div>
    </div>

    <div class="controls">
        <div class="control-row">
            <button class="arrow" onclick="changeDirection('up')">▲</button>
        </div>
        <div class="control-row">
            <button class="arrow" onclick="changeDirection('left')">◀</button>
            <button class="arrow" onclick="changeDirection('down')">▼</button>
            <button class="arrow" onclick="changeDirection('right')">▶</button>
        </div>
    </div>

    <div class="actions">
        <button class="pixel-btn" onclick="startGame()">NEW GAME</button>
        <button class="pixel-btn" onclick="togglePause()">PAUSE</button>
        <button class="pixel-btn" onclick="toggleSound()">SOUND</button>
    </div>

    <div class="keypad">
        <div class="key">1 <small>.,?!</small></div>
        <div class="key">2 <small>ABC</small></div>
        <div class="key">3 <small>DEF</small></div>
        <div class="key">4 <small>GHI</small></div>
        <div class="key">5 <small>JKL</small></div>
        <div class="key">6 <small>MNO</small></div>
        <div class="key">7 <small>PQRS</small></div>
        <div class="key">8 <small>TUV</small></div>
        <div class="key">9 <small>WXYZ</small></div>
    </div>

    <div class="phone-bottom">NOKIA CONNECTING PEOPLE</div>
    <div class="instructions">
        ARROW KEYS / WASD = MOVE<br>
        SPACE = PAUSE & RESUME<br>
        EAT • GROW • SURVIVE
    </div>
</div>

<script>
(() => {
    const canvas = document.getElementById("game");
    const ctx = canvas.getContext("2d");

    const SIZE = 16;
    const CELLS = 20;
    const TOTAL = SIZE * CELLS;

    let snake, food, direction, nextDirection;
    let score = 0;
    let best = 0;
    let running = false;
    let paused = false;
    let gameOver = false;
    let soundOn = true;
    let timer = null;
    let audio = null;

    try {
        best = Number(localStorage.getItem("nokiaSnakeBest") || 0);
    } catch(e) {}

    document.getElementById("best").textContent = best;

    function beep(freq=600, duration=0.09) {
        if (!soundOn) return;
        try {
            audio = audio || new (window.AudioContext || window.webkitAudioContext)();
            if (audio.state === "suspended") audio.resume();

            const oscillator = audio.createOscillator();
            const gain = audio.createGain();

            oscillator.type = "square";
            oscillator.frequency.value = freq;
            gain.gain.setValueAtTime(0.07, audio.currentTime);
            gain.gain.exponentialRampToValueAtTime(
                0.001, audio.currentTime + duration
            );

            oscillator.connect(gain);
            gain.connect(audio.destination);
            oscillator.start();
            oscillator.stop(audio.currentTime + duration);
        } catch(e) {}
    }

    function randomFood() {
        let f;
        do {
            f = {
                x: Math.floor(Math.random() * CELLS),
                y: Math.floor(Math.random() * CELLS)
            };
        } while (snake.some(s => s.x === f.x && s.y === f.y));
        return f;
    }

    function startGame() {
        clearInterval(timer);

        snake = [
            {x:9,y:10},
            {x:8,y:10},
            {x:7,y:10}
        ];

        direction = "right";
        nextDirection = "right";
        score = 0;
        running = true;
        paused = false;
        gameOver = false;
        food = randomFood();

        document.getElementById("score").textContent = score;
        document.getElementById("overlay").style.display = "none";
        document.getElementById("status").textContent = "PLAYING";

        draw();
        schedule();
    }

    function schedule() {
        clearInterval(timer);
        if (running && !paused) {
            const speed = Math.max(65, 155 - score * 3);
            timer = setInterval(update, speed);
        }
    }

    function changeDirection(d) {
        if (!running || paused) return;

        const opposite = {
            up:"down", down:"up",
            left:"right", right:"left"
        };

        if (d !== opposite[direction]) {
            nextDirection = d;
        }
    }

    function update() {
        if (!running || paused) return;

        direction = nextDirection;

        const head = {...snake[0]};

        if (direction === "up") head.y--;
        if (direction === "down") head.y++;
        if (direction === "left") head.x--;
        if (direction === "right") head.x++;

        const eating = head.x === food.x && head.y === food.y;

        if (
            head.x < 0 || head.x >= CELLS ||
            head.y < 0 || head.y >= CELLS
        ) {
            endGame();
            return;
        }

        const bodyToCheck = eating ? snake : snake.slice(0, -1);

        if (bodyToCheck.some(s => s.x === head.x && s.y === head.y)) {
            endGame();
            return;
        }

        snake.unshift(head);

        if (eating) {
            score += 10;
            beep(850, 0.08);
            food = randomFood();

            document.getElementById("score").textContent = score;

            if (score > best) {
                best = score;
                document.getElementById("best").textContent = best;
                try {
                    localStorage.setItem("nokiaSnakeBest", best);
                } catch(e) {}
            }

            schedule();
        } else {
            snake.pop();
        }

        draw();
    }

    function draw() {
        ctx.fillStyle = "#b9cda0";
        ctx.fillRect(0,0,TOTAL,TOTAL);

        // Subtle LCD pixel grid
        ctx.strokeStyle = "rgba(45,70,35,0.09)";
        ctx.lineWidth = 0.5;

        for (let i=0;i<=CELLS;i++) {
            ctx.beginPath();
            ctx.moveTo(i*SIZE,0);
            ctx.lineTo(i*SIZE,TOTAL);
            ctx.stroke();

            ctx.beginPath();
            ctx.moveTo(0,i*SIZE);
            ctx.lineTo(TOTAL,i*SIZE);
            ctx.stroke();
        }

        // Food
        ctx.fillStyle = "#263e20";
        ctx.fillRect(food.x*SIZE+3,food.y*SIZE+3,10,10);
        ctx.fillRect(food.x*SIZE+6,food.y*SIZE,4,4);

        // Snake
        snake.forEach((part,index) => {
            ctx.fillStyle = index === 0 ? "#172d17" : "#293e24";
            ctx.fillRect(
                part.x*SIZE+1,
                part.y*SIZE+1,
                SIZE-2,
                SIZE-2
            );

            // Pixel highlights
            ctx.fillStyle = "#b9cda0";
            if (index === 0) {
                if (direction === "right" || direction === "left") {
                    ctx.fillRect(part.x*SIZE+5,part.y*SIZE+4,2,2);
                    ctx.fillRect(part.x*SIZE+5,part.y*SIZE+10,2,2);
                } else {
                    ctx.fillRect(part.x*SIZE+4,part.y*SIZE+5,2,2);
                    ctx.fillRect(part.x*SIZE+10,part.y*SIZE+5,2,2);
                }
            }
        });
    }

    function endGame() {
        clearInterval(timer);
        running = false;
        gameOver = true;
        beep(220,0.35);

        document.getElementById("status").textContent = "GAME OVER";

        const overlay = document.getElementById("overlay");
        overlay.style.display = "flex";
        overlay.innerHTML = `
            <h2>GAME OVER</h2>
            <p>
                YOUR SCORE: ${score}<br>
                BEST SCORE: ${best}<br><br>
                ${score >= best && score > 0 ? "NEW RECORD!" : "TRY AGAIN!"}
            </p>
            <button class="pixel-btn" onclick="startGame()">
                ▶ PLAY AGAIN
            </button>
        `;
    }

    function togglePause() {
        if (!running) return;

        paused = !paused;

        if (paused) {
            clearInterval(timer);
            document.getElementById("status").textContent = "PAUSED";

            const overlay = document.getElementById("overlay");
            overlay.style.display = "flex";
            overlay.innerHTML = `
                <h2>PAUSED</h2>
                <p>TAKE A BREAK!</p>
                <button class="pixel-btn" onclick="togglePause()">
                    ▶ RESUME
                </button>
            `;
        } else {
            document.getElementById("overlay").style.display = "none";
            document.getElementById("status").textContent = "PLAYING";
            schedule();
        }
    }

    function toggleSound() {
        soundOn = !soundOn;
        document.getElementById("soundBtn").textContent =
            soundOn ? "SOUND ON" : "SOUND OFF";
        if (soundOn) beep(650,0.08);
    }

    document.addEventListener("keydown", function(e) {
        const keys = {
            ArrowUp:"up", w:"up", W:"up",
            ArrowDown:"down", s:"down", S:"down",
            ArrowLeft:"left", a:"left", A:"left",
            ArrowRight:"right", d:"right", D:"right"
        };

        if (keys[e.key]) {
            e.preventDefault();
            changeDirection(keys[e.key]);
        }

        if (e.code === "Space") {
            e.preventDefault();
            togglePause();
        }
    });

    // Initial LCD welcome screen
    snake = [
        {x:9,y:10},
        {x:8,y:10},
        {x:7,y:10}
    ];
    food = {x:14,y:10};
    direction = "right";
    nextDirection = "right";
    draw();

    // Expose functions to HTML buttons
    window.startGame = startGame;
    window.changeDirection = changeDirection;
    window.togglePause = togglePause;
    window.toggleSound = toggleSound;
})();
</script>
</body>
</html>
"""

# Escape the HTML so it can safely be embedded in an iframe attribute.
page = f"""
<iframe
    srcdoc="{html.escape(game_html, quote=True)}"
    width="100%"
    height="900"
    style="border:0; max-width:500px; display:block; margin:auto;"
    allow="autoplay"
    sandbox="allow-scripts allow-same-origin">
</iframe>
"""

display(HTML(page))
```
